```python
import torch
```

* 组合
    * torch.cat()   沿着已有的数据的某一维度进行拼接
    * torch.stack() 新增维度，按照指定的维度进行叠加
* 分块
    * torch.chunk()
    * torch.split()

```python
a = torch.Tensor([[1,2], [3,4]])
b = torch.Tensor([[5,6], [7,8]])
a, b
```

```python
# 0 2,2->4,2
# 1 2,2->2,4
torch.cat([a, b], 0), torch.cat([a, b], 1)
```

```python
# 2,2 -> 1,2,2 -> 2, 2, 2
# 2,2 -> 2,1,2 -> 2, 2, 2
# 2,2 -> 2,1,1 -> 2, 2, 2
torch.stack([a, b], 0), torch.stack([a, b], 1), torch.stack([a,b], 2)
```

```python
a = torch.Tensor([[1,2,3, 4], [4,5,6,7]])
# 2,4 -> 2, 1, 4
# 2,4 -> 2, 2, 2
torch.chunk(a, 2, 0), torch.chunk(a, 2, 1)
```

```python
# 2,3 -> 1, 2, 3
# 2,3 -> 2, 2, 2
torch.split(a, 2, 0), torch.split(a, 2, 1)
```
