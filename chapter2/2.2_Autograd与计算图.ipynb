## 2.2 Autograd与计算图
### 2.2.1 Tensor的自动求导:Autograd

```python
import torch
# 参数requires_grad=True创建支持自动求导的Tensor
a = torch.randn(2,2,requires_grad=True)
b = torch.randn(2,2)
a, b
```

```python
a.requires_grad, b.requires_grad
```

```python
# requires_grad_()将Tensor变为需要求导
b.requires_grad_()
a.requires_grad, b.requires_grad
```

```python
c = a + b
c.requires_grad
```

```python
a.grad_fn, b.grad_fn, c.grad_fn
```

```python
d = c.detach()
d.requires_grad
```

### 2.2.2 计算图

```python
import torch
x = torch.randn(1)
w = torch.ones(1, requires_grad=True)
b = torch.ones(1, requires_grad=True)
x.is_leaf, w.is_leaf, b.is_leaf
```

```python
x.requires_grad, w.requires_grad, b.requires_grad
```

```python
y = w * x
z = y + b
y.is_leaf, z.is_leaf
```

```python
y.grad_fn, z.grad_fn
```

```python
x, w, b, y, z
```
