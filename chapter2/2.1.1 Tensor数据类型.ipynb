### 2.1.1 Tensor数据类型

numpy与torch.Tensor数据类型对比
numpy   | CPU Tensor | GPU Tensor
--------|------|-------
int8    | torch.CharTensor    | torch.cuda.CharTensor
int16   | torch.ShortTensor   | torch.cuda.ShortTensor
int32   | torch.IntTensor     | torch.cuda.IntTensor
int64   | torch.LongTensor    | torch.cuda.LongTensor
uint8   | torch.ByteTensor    | torch.cuda.ByteTensor
uint16  | 
uint32  | 
uint64  | 
float16 | N/A                 | torch.cuda.HalfTensor
float32 | torch.FloatTensor   | torch.cuda.FloatTensor
float64 | torch.DoubleTensor  | torch.cuda.DoubleTensor
complex64 | 
complex128| 

```python
a = np.array([[1,2],[3,4]], dtype=np.int32)
a, a.dtype, a.shape
```

* Tensor类型转换：
    * type(new_type)
    * type_as()
    * int()/float()/double()
* Numpy类型转换：
    * astype()

```python
# 创建Tensor默认类型为torch.FloatTensor
a = torch.Tensor(2, 2)
a
```

```python
# 使用int()/float()/double()进行数据类型转换
b = a.double()
print(b)
print(a.int(), a.float(), a.double())
```

```python
# 使用type()函数进行数据类型转换
c = a.type(torch.DoubleTensor)
print(c)
```

```python
# 使用type_as()函数进行数据类型转换
# 并不需要知道具体的数据类型
d = a.type_as(b)
d
```

```python
a = np.ones((4,4))
print(a, a.dtype, a.shape)
b = a.astype(np.float32)
print(b, b.dtype, b.shape)
```
