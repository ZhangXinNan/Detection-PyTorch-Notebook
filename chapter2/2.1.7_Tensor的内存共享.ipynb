# 3 Tensor与Numpy互转
- Tensor.numpy()->numpy
- Tensor.from_numpy()->Tensor
- Tensor.tolist()->list

```python
# Tensor转Numpy
a = torch.randn(2,3,4)
print(a.shape, a.dtype, type(a))
print(a.size())
print(a)
```

```python
b = a.numpy()
print(b.shape, b.dtype, type(b))
print(b)
```

```python
# numpy转Tensor
print(type(b), b.shape, b.dtype)
c = torch.from_numpy(b)
print(type(c), c.shape, c.dtype)
```

```python
# Tensor转list
print(type(a), a.shape, a.dtype)
d = a.tolist()
print(type(d), d)
```
