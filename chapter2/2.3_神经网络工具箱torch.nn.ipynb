```python
import torch.nn.functional as F
label = torch.Tensor([0, 1, 1, 0]).long()
# 实例化nn中的交叉熵损失类
criterion = torch.nn.CrossEntropyLoss()
loss_nn = criterion(output, label)
loss_nn
```

### 2.3.3 优化器nn.optim

```python
import torch
from mlp import MLP
model = MLP(28 * 28, 300, 200, 10)
model
```

```python
# 采用SGD优化器
optimizer = torch.optim.SGD(model.parameters(), lr =0.001, momentum=0.9)
# optimizer = torch.optim.Adam([var1, var2], lr=0.0001)
data = torch.randn(10, 28*28)
output = model(data)
label = torch.Tensor([1, 0, 4, 7, 9, 3, 4, 5, 3, 2]).long()
```

```python
# 求损失
criterion = torch.nn.CrossEntropyLoss()
loss = criterion(output, label)
loss
```

```python
# 清空梯度每次优化前需要进行此操作
optimizer.zero_grad()
# 损失的反向传播
loss.backward()
# 利用 优化器进行梯度更新
optimizer.step()
```
