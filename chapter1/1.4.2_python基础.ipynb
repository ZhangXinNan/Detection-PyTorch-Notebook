# 3 高阶函数
## map()函数

````python
def f(x):
    return x * x
# python2直接返回结果；python3返回迭代器
map(f, [1,2,3,4,5])
list(map(f, [1,2,3,4,5]))
````

````python
list(map(str, range(4)))
````

````python
list(map(lambda x: x ** 2, [1, 2, 3, 4, 5]))   # 使用 lambda 匿名函数
````

## reduce()函数

````python
from functools import reduce
def f(x, y):
    return x * 10 + y
reduce(f, [1,3,5,7,9])
````

````python
# 使用 lambda 匿名函数
reduce(lambda x, y: x+y, [1,2,3])
````

## filter()函数

````python
def is_odd(n):
    return n % 2 == 0
list(filter(is_odd, [1,2,3,4,5,6]))
````

## sorted()函数

````python
sorted([3, 5, -87, 0, -21], key=abs, reverse=True)
````

````python
sorted([3, 5, -87, 0, -21], key=abs, reverse=False)
````

# 4 迭代器与生成器

list/dict/set都属于可迭代对象，可用iter()函数封装成迭代器，使用next()访问元素。

````python
x = [1,2,3]
y = iter(x)
z = iter(x)
next(y), next(y), next(z)
````

使用yield创建生成器

````python
def f():
    yield 1
    yield 2
    yield 3

f1 = f()
print([next(f1) for i in range(2)])
````

````python
def fibonacci():
    a = [1,1]
    while True:
        a.append(sum(a))
        yield a.pop(0)

for x in fibonacci():
    print(x)
    if x > 50:
        break
````

```
__iter__()返回迭代器本身
__next__()返回容器中的下一个
```

````python
class MyIter:
    def __init__(self):
        self._start = 0

    def __iter__(self):
        self._start += 8
        return self

    def __next__(self):
        if self._start < 10:
            self._start += 1
            return self._start
        else:
            raise StopIteration


a = MyIter()
for i in a:
    print(i)
````
